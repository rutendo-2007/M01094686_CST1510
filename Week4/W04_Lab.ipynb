### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```
label: R-001
value: 90.0
limit: 100.0
status: OK
```

````python
# D9
"""Printing out keys alone, values alone and then both"""
record = {"label": "R-001", "value": 90.0, "limit": 100.0, "status": "OK"}
for key in record:#auto loops through each key and prints it out
    print(key)

for value in record.values():# points to the exact value stored and printed out
    print(value)


for key, value in record.items():#prints both key and value
    print(f"{key} : {value}")
````

### D13. Read a file line by line and split each line

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.


````python
# D13
"""reading records_sample.txt line by line and spliting the line"""
with open("records_sample.txt") as f:
    for  line in f:#loops through and gets a line with\n
        line=line.strip()#strip removes the  \n so two lines that look identical are seen as  equal or no longer case sensitive
        label, value, limit = line.split(",")#spliting the line to store separate values in variables
        value = float(value)
        limit = float(limit)
        print(f"{label} {value} {limit}")

````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path):
    """reads `label,value,limit` lines from a file and returns a list of dicts"""
    records = []
    try:
        with open(path) as f:
            for line in f:
                line = line.strip()
                label, value, limit = line.split(",")#spliting the line to store separate values in variables
                value = float(value)
                limit = float(limit)
                records.append({"label": label, "value": value, "limit": limit})
    except FileNotFoundError:
        print(f"File {path} not found.")
        return []
    return records

records = load_records("records_sample.txt")


for rec in records:
    if rec["value"] > rec["limit"]:
        print(f"{rec['label']} : OVER LIMIT")
    else:
        print(f"{rec['label']} : OK")
        
file1 = load_records("records_sample.txt")
file2 = load_records("nope.txt")
print(f"{file1}\n {file2}")
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2
"""Writing results to results.txt file"""
with open("results.txt", "w") as f:
    for record in records:#looping through Each Record
        if record["value"] > record["limit"]:
            f.write(f"{record["label"]},OVER LIMIT\n ")#{record["label"]} gets value stored in the key label and is written into the file
        else:
            f.write(f"{record["label"]},OK \n")

with open("results.txt", "r") as f:
    for line in f:
        print(line.strip())#printing the results line by line and removing the \n
    
````
