### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90):#defining function status_of that takes 'percent' and an optional parameter 'warning_at' with a default value of 90.
   """ Returns a status based on value of 'percent'"""
   if percent >= 100:
        return "OVER LIMIT" 
   elif percent >= warning_at:
        return "WARNING"
   else:
        return "OK"


print(status_of(95))
print(status_of(85))
print(status_of(85,80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

Parameters(found when defining a function)
Arguments (found when calling a function)
default or positional arguments/parameters
*args = tp arbitary arguments(do no know how many arguments)unknown
**kwargs = keyword arbitary arguments(can add as many arguments the user wishes)infinite
kwargs can be used for registration forms or shopping or entering many details 

````python
# D8
def total(*numbers):
    """returns the sum of all the numbers passed in"""
    return sum(numbers)


def configure(**settings):
    """loops over `settings.items()` and prints each one as `key = value`"""
    for key, value in settings.items():#looping over each item(number determined by arguments passed) within the parameter(dictionary) settings 
        print(f"{key} = {value}")

print(total(1,2,3))
print(total(10,20,30,40))
print(total())


configure(host="10.0.0.1",port=8080)
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and

````python
# D12
def check_record(value,limit):
    """returns both difference and percent`"""
    difference = value - limit
    percent = ((value/limit)* 100)
    return difference, percent#returning both difference and percent in the same line as a tuple as return cannot be written twice in the same function definition
    

difference,percent = check_record(87.0,100.0)#calling difference and percent as a tuple, split into 2 variables
print(f"Difference: {difference} \nPercent: {percent}" )
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values,limit=100):
    """Returns how many of the values are over the limit."""
    count = 0
    for value in values:#for each value in the values tuple, we check if it is greater than the limit and increment count by one if it is
        if value > limit:#eg if 87 >100 count not incremented
            count += 1
    return count

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101,limit=50))
print(count_over())
help(count_over)
````
